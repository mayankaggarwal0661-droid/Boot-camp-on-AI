```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("=== ML Model Trainer: Student Pass Prediction ===")

n = int(input("Enter number of student records to generate for training: "))
test_percent = float(input("Enter test data percentage (e.g. 20 for 20%): "))
test_ratio = test_percent / 100
seed = int(input("Enter a random seed (any number, changes the data generated): "))

np.random.seed(seed)

study_hours = np.random.uniform(0, 10, n)
attendance = np.random.uniform(40, 100, n)
passed = ((study_hours * 5 + attendance * 0.5 + np.random.normal(0, 10, n)) > 60).astype(int)

df = pd.DataFrame({
    'study_hours': study_hours,
    'attendance': attendance,
    'passed': passed
})
print("\nSample of generated training data:")
print(df.head())

X = df[['study_hours', 'attendance']]
y = df['passed']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_ratio, random_state=seed)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression()
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
print("\nModel Accuracy:", round(accuracy, 3))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred, labels=[0, 1]))
print("Classification Report:")
print(classification_report(y_test, y_pred, labels=[0, 1], zero_division=0))

print("\n=== Try Your Own Prediction ===")
user_hours = float(input("Enter study hours (0-10): "))
user_attendance = float(input("Enter attendance percentage (40-100): "))

new_student = pd.DataFrame({'study_hours': [user_hours], 'attendance': [user_attendance]})
new_student_scaled = scaler.transform(new_student)
prediction = model.predict(new_student_scaled)
probability = model.predict_proba(new_student_scaled)

print(f"\nPrediction: {'Pass' if prediction[0] == 1 else 'Fail'}")
print(f"Confidence: {round(max(probability[0]) * 100, 2)}%")
```
